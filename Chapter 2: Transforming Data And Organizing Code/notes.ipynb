# 2.7 Forward Declarations and Definitions

- Forward Declaration:
    - tells the compiler about the existence of a function before we define the function's body
    - used to tell the compiler about the existence of some function that has been 
    - used with other identifiers in C++, variables, types


- Function Declaration (function prototype):
    - consists of the function's return type, name, parameter type, terminated with a semicolon
    - does not need to specify the name of the parameter



Difference between declaration vs definition:
- declaration: tells the compiler about the existence of an identifier and its associated type information

// Declaration: says this function exists elsewhere
int add(int, int);

- definition: a declaration that actually implements or instantiates the identifier

// Definition: provides its implementation
int add(int a, int b) {
  return a + b;
}


## One Definition Rule:

- in a file, each function, variable, type, or template in a scope can only have 1 definition
    causes the compiler to issue a redefinition error

int add(int x, int y)
{
     return x + y;
}

int add(int x, int y) // violation of ODR, we've already defined function add(int, int)
{
     return x + y;
}

int main()
{
    int x{};
    int x{ 5 }; // violation of ODR, we've already defined x
}


- in a program, each function in a given scope can only have 1 definition
    cause the linker to issue a redefinition  error

- types, templates, inline functions, and inline variables are allowed to have duplicate definitions in different files, so long as each definition is identical.
    cause undefined behavior

What is f(x) prototype?

- Function prototype is a declaration statement that includes a f(x) name, return type, parameter type, and optionally parameter name
- does not include the f(x) body
- tells the compiler about the existence of a function before it is defined


What is a forward declaration?

- a forward declaration tells the compiler an identifier exists before it is actually defined

# 2.10 Introduction To Preprocessors

- preprocessing phase
    - makes various changes to the text of the code file
    - happens in memory or using temporary files
    - strips out comments, ensures each code file ends in newline
    - processes #include directives
    - translation unit - when the processor finished processing code file. TU is what is compiled by the compiler


- translation
    - process of preprocessing, compiling, and linking


- preprocessing directives
    - scans through code files from top to bottom looking for preprocessor directives
    - preprocessor directives start with a #, end on a newline

## Macros

- macro defines
    - preprocessor instructions that run before the compiler sees your code
    - #define directive used to create a macro
    - macro = a rule that defines how input text is converted into replacement output text. There are 2 types of macros:
        - object-like 
            - defined in one of 2 ways:
                #define IDENTIFIER
                #define IDENTIFIER substitution_text
        - function-like
            - generally unsafe, can be done by normal function

## Conditional Compilation

- conditional compilation preprocessor directive allow you to specify if something will or won't compile
- #ifdef, #ifndef, #endif
